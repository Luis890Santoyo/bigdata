## Actividad 3 — Estadística descriptiva

**Propósito:** aplicar medidas estadísticas básicas a una variable poblacional.

Se selecciona el año **2020** (dentro del periodo histórico, 1951–2023) y se filtran los datos correspondientes. Sobre la variable `Annual change in population` se calculan media, mediana, varianza, desviación estándar, mínimo y máximo.

```python
df_2020 = df[df["Year"] == 2020]
cambio_poblacional = df_2020["Annual change in population"].dropna()
cambio_poblacional.head()
```

```python
media = cambio_poblacional.mean()
mediana = cambio_poblacional.median()
varianza = cambio_poblacional.var()
desviacion_estandar = cambio_poblacional.std()
minimo = cambio_poblacional.min()
maximo = cambio_poblacional.max()

print("Media:", media)
print("Mediana:", mediana)
print("Varianza:", varianza)
print("Desviación estándar:", desviacion_estandar)
print("Mínimo:", minimo)
print("Máximo:", maximo)
```

**Entidades con el valor mínimo y máximo**

```python
entidad_min = df_2020.loc[cambio_poblacional.idxmin(), "Entity"]
entidad_max = df_2020.loc[cambio_poblacional.idxmax(), "Entity"]

print("Entidad con el valor mínimo:", entidad_min, "->", minimo)
print("Entidad con el valor máximo:", entidad_max, "->", maximo)
```

1. **¿Cuál es el valor de la media?**

4. **¿Qué entidad presenta el valor mínimo y cuál el máximo?**
   El valor mínimo (-494,016) corresponde a Venezuela. El valor máximo (75,707,590) corresponde a "World", que es una entidad agregada (el total mundial) y no un país individual; entre los países, el mayor crecimiento en 2020 lo tuvo India.
